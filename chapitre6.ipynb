```python
import pandas as pd 

df = pd.read_csv("appartements.csv")

print(df.isna().sum())
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()
df = df.dropna(subset=["prix"])
df["etage"] = df["etage"].fillna(df["etage"].median())
df["etage"] = df["etage"].astype(int)
df = df.reset_index(drop = True)
print(df)
```

```python
df["prix_m2"] = (df["prix"] / df["surface"]).round(2)
df.loc[df["prix_m2"].idxmax()]
```

```python
df.to_csv("appartements_propre.csv", index = False)
```
