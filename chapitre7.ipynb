```python
import pandas as pd 

df = pd.read_csv("appartements_propre.csv")

print(df.nlargest(3,"prix"))
print(df.sort_values(["quartier","surface"]))
```

```python
print(df["pieces"].value_counts().sort_index())
```

```python
print(df.groupby("pieces")["surface"].mean().round(1))
```

```python
df.groupby("quartier").agg(
    nb = ("id","count"),
    prix_m2_moy = ("prix_m2","mean"),
    prix_m2_min = ("prix_m2", "min"),
).round(2).sort_values("prix_m2_moy",ascending = False)
```
