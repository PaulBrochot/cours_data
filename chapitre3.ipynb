```python
import numpy as np 

surfaces = np.array([30, 35, 45, 50, 60, 70, 80, 90, 100, 120])
prix = np.array([150, 170, 210, 240, 275, 320, 360, 410, 445, 530])

np.polyfit(surfaces, prix, 1)

def erreur_moyenne(y_vrai,y_pred):
    y_vrai = np.array(y_vrai)
    y_pred = np.array(y_pred)
    return np.abs(y_pred - y_vrai).mean()

erreur_moyenne([300, 400], [320, 390]) 
```

```python
def predict_moyenne (X, moyenne):
    return np.full(len(X), moyenne)


predict_moyenne([30, 75, 120], 300)   # [300 300 300]
```

```python
def separer(X, y, ratio, graine): 
    X, y = np.asarray(X), np.asarray(y)
    rng = np.random.default_rng(graine)
    indices = rng.permutation(len(X))
    coupure = int(len(X) * ratio)
    train, test = indices[:coupure], indices[coupure:]
    return X[train], y[train], X[test], y[test]

X_train, y_train, X_test, y_test = separer(surfaces, prix, 0.8, 42)
print(len(X_train), len(X_test))
print(X_test, y_test)

```

```python
class RegressionLineaireNP:

    def fit(self, X, y):
        X, y = np.asarray(X), np.asarray(y)
        x_moyen = X.mean()
        y_moyen = y.mean()

        numerateur = ((X - x_moyen) * (y - y_moyen)).sum()
        denominateur = ((X - x_moyen) ** 2).sum()

        self.a = numerateur / denominateur
        self.b = y_moyen - self.a * x_moyen

    def predict(self, X):
        X = np.asarray(X)
        return self.a * X + self.b

modele = RegressionLineaireNP()
modele.fit(surfaces, prix)
print(modele.a, modele.b)            # ≈ 4.2407  22.6312
print(np.polyfit(surfaces, prix, 1)) # pour comparer
print(erreur_moyenne(prix, modele.predict(surfaces)))  # ≈ 2.34
```
