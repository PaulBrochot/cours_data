```python
import numpy as np
surfaces = np.array([30, 35, 45, 50, 60, 70, 80, 90, 100, 120])
prix = np.array([150, 170, 210, 240, 275, 320, 360, 410, 445, 530])

print(prix[-3:])
print(surfaces[prix.argmax()])
print(surfaces[prix>300])
print(len(surfaces[(surfaces >= 50) & (surfaces <= 90)]))
prix_m2 = prix / surfaces
print(prix_m2)
print(prix / surfaces)
print(round(prix_m2.mean(),2))
```
